# Count entries for each experiment

```python
import pandas as pd

df = pd.read_csv('EnzEngDB_V1/meta_data.csv', encoding='latin-1')
df
```

```python
# Read in each of the experiments and write in how many entries are in there
num_proteins, num_rows, num_fit, num_rxns = [], [], [], []
all_df = pd.DataFrame()
for exp in df['experiment_id'].values:
    exp_df = pd.read_csv(f'EnzEngDB_V1/experiments/{exp}.csv')
    exp_df = exp_df.dropna(subset='fitness_value')
    unique_proteins = len(set(list(exp_df['aa_sequence'].values)))
    unique_reactions = len(set(list(exp_df['smiles_string'].values)))
    unique_fitness = len(set(list(exp_df['fitness_value'].values)))
    print(exp, unique_fitness)
    num_proteins.append(unique_proteins)
    num_rows.append(len(exp_df))
    num_fit.append(unique_fitness)
    num_rxns.append(unique_reactions)
    all_df = pd.concat([all_df, exp_df])

df['num_proteins'] = num_proteins
df['num_rows'] = num_rows
df['num_fitness_entries'] = num_fit
df['num_rxns'] = num_rxns
```

```python
df.to_csv('output/meta_data_numbers_V3.csv', index=False)
```

```python
# Also convert all of them to INCHI strings
from rdkit.Chem import inchi
from rdkit import Chem

inchi_substrates = []
inchi_products = []
# Just do it for the substrates and products
for substrates, products in df[['substrate_smiles', 'product_smiles']].values:
    inchi_subs = []
    inchi_prods = []
    if isinstance(substrates, str):
        for substrate in substrates.split('.'):
            mol = Chem.MolFromSmiles(substrate)
            inchi_string = inchi.MolToInchi(mol)
            inchi_subs.append(inchi_string)
    if isinstance(products, str):
        for product in products.split('.'):
            mol = Chem.MolFromSmiles(product)
            inchi_string = inchi.MolToInchi(mol)
            inchi_prods.append(inchi_string)
    inchi_substrates.append(' '.join(inchi_subs))
    inchi_products.append(' '.join(inchi_prods))
df['INCHI substrates'] = inchi_substrates
df['INCHI products'] = inchi_products
df.to_csv('output/meta_data_numbers_inchi_V3.csv', index=False)
```
